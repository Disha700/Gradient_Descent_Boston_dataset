```python
import numpy as np
```

```python
df_train= np.genfromtxt(r"C:\Users\Dell\Downloads\0000000000002417_training_boston_x_y_train.csv", delimiter=',')
df_test= np.genfromtxt(r"C:\Users\Dell\Downloads\0000000000002417_test_boston_x_test.csv", delimiter=',')
```

```python
X_train = df_train[:, :-1]
Y_train = df_train[:, -1]

X_test= df_test
```

```python
#Feature scaling
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std
```

```python
X_train = np.c_[np.ones(X_train.shape[0]), X_train]
X_test = np.c_[np.ones(X_test.shape[0]), X_test]
```

```python
def cost(points, Y, m):

    total_cost = 0
    M = len(points)

    for i in range(M):

        y_pred = 0

        # Calculate prediction for all features
        for j in range(len(m)):
            y_pred += m[j] * points[i, j]

        total_cost += (1/M) * ((Y[i] - y_pred) ** 2)

    return total_cost
```

```python
def step_gradient(points, Y, learning_rate, m):

    M = len(points)

    # One slope for each feature
    m_slope = [0] * len(m)

    for i in range(M):

        # Calculate predicted Y
        y_pred = 0

        for j in range(len(m)):
            y_pred += m[j] * points[i, j]

        error = Y[i] - y_pred

        # Calculate gradient for every feature
        for j in range(len(m)):
            m_slope[j] += (-2/M) * error * points[i, j]

    # Update all slopes
    new_m = [0] * len(m)

    for j in range(len(m)):
        new_m[j] = m[j] - learning_rate * m_slope[j]

    return new_m
```

```python
def gd(points, Y, learning_rate, num_iterations):

    # Number of parameters = number of features
    # + 1 intercept because we added a column of 1s
    m = [0] * points.shape[1]

    for i in range(num_iterations):

        m = step_gradient(
            points,
            Y,
            learning_rate,
            m
        )

        # Print cost occasionally
        if i % 1000 == 0:
            print(
                i,
                "cost",
                cost(points, Y, m)
            )

    return m
```

```python
learning_rates = [0.001, 0.01, 0.05, 0.1]
iterations = [1000, 5000, 10000]

best_m = None
best_cost = float("inf")
best_lr = None
best_iterations = None

for learning_rate in learning_rates:

    for num_iterations in iterations:

        print(
            "\nLearning Rate:",
            learning_rate,
            "Iterations:",
            num_iterations
        )

        m = gd(
            X_train,
            Y_train,
            learning_rate,
            num_iterations
        )

        current_cost = cost(
            X_train,
            Y_train,
            m
        )

        print("Final Cost:", current_cost)

        if current_cost < best_cost:

            best_cost = current_cost
            best_m = m
            best_lr = learning_rate
            best_iterations = num_iterations
print("\nBest Learning Rate:", best_lr)
print("Best Iterations:", best_iterations)
print("Best Cost:", best_cost)
```

```python
predictions = []

for i in range(len(X_test)):

    y_pred = 0

    for j in range(len(best_m)):
        y_pred += best_m[j] * X_test[i, j]

    predictions.append(y_pred)


# Convert predictions to NumPy array
predictions = np.array(predictions)


np.savetxt(
    "boston_predictions.csv",
    predictions,
    delimiter=",",
    fmt="%.5f"
)

print("\nPredictions saved successfully!")
```
